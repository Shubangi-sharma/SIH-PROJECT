```python
import pandas as pd

# =========================
# REGION BOUNDING BOXES
# =========================

# North Region
NORTH_LAT_MIN = 23.015321
NORTH_LAT_MAX = 35.603247
NORTH_LON_MIN = 69.2328816148548
NORTH_LON_MAX = 79.473024292358

# West Region
WEST_LAT_MIN = 14.90536962
WEST_LAT_MAX = 24.51767769
WEST_LON_MIN = 68.501592
WEST_LON_MAX = 80.56554985


# =========================
# LOAD FIRMS DATA
# =========================

firms = pd.read_csv("/content/drive/MyDrive/DataSetForpython/fire(1).csv")

print("Total FIRMS records:", len(firms))
print("Columns:", list(firms.columns))


# =========================
# FILTER NORTH
# =========================

north_firms = firms[
    (firms["LATITUDE"] >= NORTH_LAT_MIN) &
    (firms["LATITUDE"] <= NORTH_LAT_MAX) &
    (firms["LONGITUDE"] >= NORTH_LON_MIN) &
    (firms["LONGITUDE"] <= NORTH_LON_MAX)
].copy()


# =========================
# FILTER WEST
# =========================

west_firms = firms[
    (firms["LATITUDE"] >= WEST_LAT_MIN) &
    (firms["LATITUDE"] <= WEST_LAT_MAX) &
    (firms["LONGITUDE"] >= WEST_LON_MIN) &
    (firms["LONGITUDE"] <= WEST_LON_MAX)
].copy()


# =========================
# SAVE RESULTS
# =========================

north_firms.to_csv("north_firms.csv", index=False)
west_firms.to_csv("west_firms.csv", index=False)


# =========================
# BASIC VERIFICATION
# =========================

print("\n========== RESULTS ==========")

print("\nNORTH REGION")
print("Records:", len(north_firms))
print("Date:", north_firms["ACQ_DATE"].min(), "to", north_firms["ACQ_DATE"].max())

print("\nWEST REGION")
print("Records:", len(west_firms))
print("Date:", west_firms["ACQ_DATE"].min(), "to", west_firms["ACQ_DATE"].max())

print("\nFiles created:")
print("north_firms.csv")
print("west_firms.csv")
```

```python
from sklearn.cluster import DBSCAN
```

```python
coords = firms_master[["LATITUDE", "LONGITUDE"]].to_numpy()

# Convert degrees → radians
coords_rad = np.radians(coords)

print("Total points:", len(coords_rad))
```

```python
# Earth's radius in metres
EARTH_RADIUS_M = 6_371_000

# 500 metre neighbourhood
EPS_METERS = 500
EPS_RAD = EPS_METERS / EARTH_RADIUS_M

dbscan = DBSCAN(
    eps=EPS_RAD,
    min_samples=2,
    metric="haversine",
    algorithm="ball_tree",
    n_jobs=-1
)

labels = dbscan.fit_predict(coords_rad)

firms_master["HOTSPOT_ID"] = labels

print("DBSCAN completed!")
print("Total clusters:", len(set(labels)) - (1 if -1 in labels else 0))
print("Noise points:", (labels == -1).sum())
```

```python
# Convert acquisition date to proper date format
firms_master["ACQ_DATE"] = pd.to_datetime(
    firms_master["ACQ_DATE"]
)

# Ignore DBSCAN noise
clustered = firms_master[
    firms_master["HOTSPOT_ID"] != -1
].copy()

# Calculate persistence statistics
hotspot_stats = clustered.groupby("HOTSPOT_ID").agg(
    total_detections=("HOTSPOT_ID", "size"),
    unique_days=("ACQ_DATE", "nunique"),
    first_detection=("ACQ_DATE", "min"),
    last_detection=("ACQ_DATE", "max"),
    mean_FRP=("FRP", "mean"),
    max_FRP=("FRP", "max")
).reset_index()

print("Candidate hotspots:", len(hotspot_stats))
print(hotspot_stats.head())
```

```python
from sklearn.neighbors import BallTree
import numpy as np

# ------------------------------------------------
# Prepare hotspot coordinates
# ------------------------------------------------

hotspot_coords = np.radians(
    persistent_candidates_clean[
        ["hotspot_latitude", "hotspot_longitude"]
    ].to_numpy()
)

# ------------------------------------------------
# Earth radius
# ------------------------------------------------

EARTH_RADIUS_KM = 6371.0088

# ------------------------------------------------
# OSM coordinates
# ------------------------------------------------

osm_coords = np.radians(
    osm_master[
        ["latitude", "longitude"]
    ].to_numpy()
)

# Build spatial index
osm_tree = BallTree(
    osm_coords,
    metric="haversine"
)

print("OSM spatial index created!")
print("OSM features indexed:", len(osm_master))
print("Hotspots to match:", len(hotspot_coords))
```

```python
distances, indices = osm_tree.query(
    hotspot_coords,
    k=1
)

nearest_distance_km = (
    distances[:, 0] * EARTH_RADIUS_KM
)

nearest_index = indices[:, 0]

print(nearest_distance_km[:10])
```

```python
# Categories that matter for our classification
categories = [
    "industrial",
    "power",
    "mining",
    "fuel_storage",
    "agriculture",
    "transport"
]

# Create a copy of hotspot data
hotspot_osm = persistent_candidates_clean.copy()

for category in categories:

    # Select OSM features of this category
    osm_cat = osm_master[
        osm_master["osm_category"] == category
    ].copy()

    # If category doesn't exist, fill with NaN
    if len(osm_cat) == 0:
        hotspot_osm[f"distance_to_{category}_km"] = np.nan
        continue

    # Coordinates in radians
    osm_cat_coords = np.radians(
        osm_cat[["latitude", "longitude"]].to_numpy()
    )

    # Build category-specific BallTree
    tree = BallTree(
        osm_cat_coords,
        metric="haversine"
    )

    # Find nearest feature for every hotspot
    distances, indices = tree.query(
        hotspot_coords,
        k=1
    )

    # Convert radians → km
    distances_km = (
        distances[:, 0] * EARTH_RADIUS_KM
    )

    # Store distance
    hotspot_osm[
        f"distance_to_{category}_km"
    ] = distances_km

print("OSM distance features created!")

print(
    hotspot_osm[
        [
            "HOTSPOT_ID",
            "distance_to_industrial_km",
            "distance_to_power_km",
            "distance_to_mining_km",
            "distance_to_fuel_storage_km",
            "distance_to_agriculture_km",
            "distance_to_transport_km"
        ]
    ].head(10)
)
```

```python
# Create proximity indicators

hotspot_osm["near_industrial_500m"] = (
    hotspot_osm["distance_to_industrial_km"] <= 0.5
).astype(int)

hotspot_osm["near_power_1km"] = (
    hotspot_osm["distance_to_power_km"] <= 1.0
).astype(int)

hotspot_osm["near_mining_5km"] = (
    hotspot_osm["distance_to_mining_km"] <= 5.0
).astype(int)

hotspot_osm["near_fuel_storage_2km"] = (
    hotspot_osm["distance_to_fuel_storage_km"] <= 2.0
).astype(int)

hotspot_osm["near_agriculture_1km"] = (
    hotspot_osm["distance_to_agriculture_km"] <= 1.0
).astype(int)

hotspot_osm["near_transport_1km"] = (
    hotspot_osm["distance_to_transport_km"] <= 1.0
).astype(int)

print(
    hotspot_osm[
        [
            "HOTSPOT_ID",
            "near_industrial_500m",
            "near_power_1km",
            "near_mining_5km",
            "near_fuel_storage_2km",
            "near_agriculture_1km",
            "near_transport_1km"
        ]
    ].head(10)
)
```

```python
from sklearn.neighbors import BallTree
import numpy as np

# --------------------------------------------------
# 1. Create the three hotspot groups
# --------------------------------------------------

hotspots = persistent_candidates_clean.copy()

north_mask = (
    (hotspots["hotspot_latitude"] >= NORTH_LAT_MIN) &
    (hotspots["hotspot_latitude"] <= NORTH_LAT_MAX) &
    (hotspots["hotspot_longitude"] >= NORTH_LON_MIN) &
    (hotspots["hotspot_longitude"] <= NORTH_LON_MAX)
)

west_mask = (
    (hotspots["hotspot_latitude"] >= WEST_LAT_MIN) &
    (hotspots["hotspot_latitude"] <= WEST_LAT_MAX) &
    (hotspots["hotspot_longitude"] >= WEST_LON_MIN) &
    (hotspots["hotspot_longitude"] <= WEST_LON_MAX)
)

north_only = north_mask & ~west_mask
west_only = west_mask & ~north_mask
overlap = north_mask & west_mask

print("North only:", north_only.sum())
print("West only:", west_only.sum())
print("Overlap:", overlap.sum())
print("Total:", north_only.sum() + west_only.sum() + overlap.sum())


# --------------------------------------------------
# 2. Function to calculate nearest OSM distance
# --------------------------------------------------

EARTH_RADIUS_KM = 6371.0088

def nearest_category_distance(hotspot_df, osm_df, category):

    result = np.full(len(hotspot_df), np.nan)

    osm_cat = osm_df[
        osm_df["osm_category"] == category
    ]

    if len(hotspot_df) == 0 or len(osm_cat) == 0:
        return result

    hotspot_coords = np.radians(
        hotspot_df[
            ["hotspot_latitude", "hotspot_longitude"]
        ].to_numpy()
    )

    osm_coords = np.radians(
        osm_cat[
            ["latitude", "longitude"]
        ].to_numpy()
    )

    tree = BallTree(
        osm_coords,
        metric="haversine"
    )

    distances, _ = tree.query(
        hotspot_coords,
        k=1
    )

    result = distances[:, 0] * EARTH_RADIUS_KM

    return result


# --------------------------------------------------
# 3. Calculate distances using CORRECT OSM coverage
# --------------------------------------------------

osm_features = [
    "industrial",
    "power",
    "mining",
    "fuel_storage",
    "agriculture",
    "transport"
]

hotspot_osm_correct = hotspots.copy()

for category in osm_features:

    column = f"distance_to_{category}_km"

    # Start with everything unavailable
    hotspot_osm_correct[column] = np.nan

    # ----------------------------------------------
    # NORTH ONLY → North OSM
    # ----------------------------------------------

    north_hotspots = hotspots[north_only]

    hotspot_osm_correct.loc[
        north_only, column
    ] = nearest_category_distance(
        north_hotspots,
        north_osm_std,
        category
    )

    # ----------------------------------------------
    # WEST ONLY → West OSM
    # ----------------------------------------------

    west_hotspots = hotspots[west_only]

    hotspot_osm_correct.loc[
        west_only, column
    ] = nearest_category_distance(
        west_hotspots,
        west_osm_std,
        category
    )

    # ----------------------------------------------
    # OVERLAP → BOTH OSM datasets
    # ----------------------------------------------

    overlap_osm = pd.concat(
        [north_osm_std, west_osm_std],
        ignore_index=True
    )

    overlap_hotspots = hotspots[overlap]

    hotspot_osm_correct.loc[
        overlap, column
    ] = nearest_category_distance(
        overlap_hotspots,
        overlap_osm,
        category
    )


print("\nCorrected OSM distances created!")

distance_cols = [
    "distance_to_industrial_km",
    "distance_to_power_km",
    "distance_to_mining_km",
    "distance_to_fuel_storage_km",
    "distance_to_agriculture_km",
    "distance_to_transport_km"
]

print(
    hotspot_osm_correct[distance_cols].describe().T
)
```

```python
def assign_label(row):

    industrial_score = 0
    agriculture_score = 0
    mining_score = 0

    # -----------------------------
    # INDUSTRIAL EVIDENCE
    # -----------------------------

    if pd.notna(row["distance_to_industrial_km"]):
        if row["distance_to_industrial_km"] <= 0.5:
            industrial_score += 3
        elif row["distance_to_industrial_km"] <= 2:
            industrial_score += 1

    if pd.notna(row["distance_to_power_km"]):
        if row["distance_to_power_km"] <= 1:
            industrial_score += 1

    if pd.notna(row["distance_to_fuel_storage_km"]):
        if row["distance_to_fuel_storage_km"] <= 2:
            industrial_score += 2

    if row["lc_built_ratio"] >= 0.5:
        industrial_score += 2


    # -----------------------------
    # AGRICULTURAL / VEGETATION
    # -----------------------------

    if row["lc_crops_ratio"] >= 0.5:
        agriculture_score += 3

    elif row["lc_crops_ratio"] >= 0.3:
        agriculture_score += 2

    if row["lc_grass_ratio"] >= 0.4:
        agriculture_score += 1

    if row["lc_trees_ratio"] >= 0.5:
        agriculture_score += 1

    if pd.notna(row["distance_to_agriculture_km"]):
        if row["distance_to_agriculture_km"] <= 1:
            agriculture_score += 2


    # -----------------------------
    # MINING
    # -----------------------------

    if pd.notna(row["distance_to_mining_km"]):

        if row["distance_to_mining_km"] <= 2:
            mining_score += 3

        elif row["distance_to_mining_km"] <= 5:
            mining_score += 2

    if row["lc_bare_ratio"] >= 0.5:
        mining_score += 1


    # -----------------------------
    # SELECT CLASS
    # -----------------------------

    scores = {
        "Agricultural_Vegetation": agriculture_score,
        "Industrial": industrial_score,
        "Mining_Extraction": mining_score
    }

    best_class = max(scores, key=scores.get)
    best_score = scores[best_class]

    # Require reasonable evidence
    if best_score < 3:
        return "Other_Persistent_Thermal_Source"

    # If two classes are too close, don't force a label
    sorted_scores = sorted(
        scores.values(),
        reverse=True
    )

    if len(sorted_scores) >= 2:
        if sorted_scores[0] - sorted_scores[1] < 2:
            return "Needs_Review"

    return best_class
```

```python
import pandas as pd
import numpy as np

# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv("/content/drive/MyDrive/DataSetForpython/weather.csv")

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Columns we expect to use for labeling
label_columns = [
    "distance_to_industrial_km",
    "distance_to_power_km",
    "distance_to_mining_km",
    "distance_to_fuel_storage_km",
    "distance_to_agriculture_km",
    "dominant_land_cover",
    "lc_crops_ratio",
    "lc_built_ratio",
    "lc_trees_ratio",
    "lc_grass_ratio",
    "lc_shrub_and_scrub_ratio",
    "lc_bare_ratio",
    "unique_days",
    "unique_months",
    "mean_frp",
    "max_frp"
]

for col in label_columns:
    if col in df.columns:
        print(f"✓ {col}")
    else:
        print(f"✗ MISSING: {col}")
```

```python
# ============================================================
# INDUSTRIAL SCORE
# ============================================================

df["industrial_score"] = 0.0

# Industrial proximity
df.loc[
    df["distance_to_industrial_km"] <= 0.5,
    "industrial_score"
] += 0.45

df.loc[
    (df["distance_to_industrial_km"] > 0.5) &
    (df["distance_to_industrial_km"] <= 2),
    "industrial_score"
] += 0.25


# Power infrastructure
df.loc[
    df["distance_to_power_km"] <= 1,
    "industrial_score"
] += 0.20


# Fuel/storage infrastructure
df.loc[
    df["distance_to_fuel_storage_km"] <= 2,
    "industrial_score"
] += 0.20


# Built-up land cover
df.loc[
    df["dominant_land_cover"] == "built",
    "industrial_score"
] += 0.15


# Cap score
df["industrial_score"] = df["industrial_score"].clip(upper=1.0)
```

```python
# ============================================================
# AGRICULTURE / VEGETATION SCORE
# ============================================================

df["agriculture_score"] = 0.0


# Crops dominant
df.loc[
    df["dominant_land_cover"] == "crops",
    "agriculture_score"
] += 0.45


# Crops ratio
df.loc[
    df["lc_crops_ratio"] >= 0.70,
    "agriculture_score"
] += 0.30

df.loc[
    (df["lc_crops_ratio"] >= 0.40) &
    (df["lc_crops_ratio"] < 0.70),
    "agriculture_score"
] += 0.20


# Agriculture OSM proximity
df.loc[
    df["distance_to_agriculture_km"] <= 1,
    "agriculture_score"
] += 0.25


# Vegetation context
df.loc[
    df["dominant_land_cover"].isin([
        "grass",
        "trees",
        "shrub_and_scrub"
    ]),
    "agriculture_score"
] += 0.15


# Cap
df["agriculture_score"] = df["agriculture_score"].clip(upper=1.0)
```

```python
# ============================================================
# MINING / EXTRACTION SCORE
# ============================================================

df["mining_score"] = 0.0


# Very close mining/quarry feature
df.loc[
    df["distance_to_mining_km"] <= 1,
    "mining_score"
] += 0.55


# Somewhat close
df.loc[
    (df["distance_to_mining_km"] > 1) &
    (df["distance_to_mining_km"] <= 5),
    "mining_score"
] += 0.30


# Bare land
df.loc[
    df["dominant_land_cover"] == "bare",
    "mining_score"
] += 0.30


# Shrubland can support extraction context
df.loc[
    df["dominant_land_cover"] == "shrub_and_scrub",
    "mining_score"
] += 0.10


# Cap
df["mining_score"] = df["mining_score"].clip(upper=1.0)
```

```python
# ============================================================
# DETERMINE STRONGEST EVIDENCE
# ============================================================

score_cols = [
    "industrial_score",
    "agriculture_score",
    "mining_score"
]

df["best_score"] = df[score_cols].max(axis=1)

df["second_best_score"] = (
    df[score_cols]
    .apply(lambda row: sorted(row, reverse=True)[1], axis=1)
)

df["score_margin"] = (
    df["best_score"] - df["second_best_score"]
)


def get_best_class(row):

    scores = {
        "Industrial": row["industrial_score"],
        "Agricultural_Vegetation": row["agriculture_score"],
        "Mining_Extraction": row["mining_score"]
    }

    return max(scores, key=scores.get)


df["evidence_class"] = df.apply(get_best_class, axis=1)
```

```python
# ============================================================
# CONFIDENCE
# ============================================================

def confidence(row):

    score = row["best_score"]
    margin = row["score_margin"]

    if score >= 0.70 and margin >= 0.25:
        return "High"

    elif score >= 0.50 and margin >= 0.15:
        return "Medium"

    else:
        return "Low"


df["label_confidence"] = df.apply(confidence, axis=1)
```

```python
# ============================================================
# PROVISIONAL LABEL
# ============================================================

def assign_label(row):

    if row["label_confidence"] == "Low":
        return "Needs_Review"

    return row["evidence_class"]


df["provisional_label"] = df.apply(assign_label, axis=1)
```

```python
print("\n========== LABEL DISTRIBUTION ==========\n")

print(
    df["provisional_label"]
    .value_counts()
)

print("\n========== CONFIDENCE ==========\n")

print(
    pd.crosstab(
        df["provisional_label"],
        df["label_confidence"]
    )
)
```

```python
# ============================================================
# ADD OTHER PERSISTENT THERMAL SOURCE
# ============================================================

def final_provisional_label(row):

    # Strong enough evidence
    if row["best_score"] >= 0.50:

        # But if competition between classes is too close,
        # send it for review
        if row["score_margin"] < 0.15:
            return "Needs_Review"

        return row["evidence_class"]

    # No strong contextual evidence
    return "Other_Persistent_Thermal_Source"


df["provisional_label"] = df.apply(
    final_provisional_label,
    axis=1
)
```

```python
# ============================================================
# FINAL PROVISIONAL LABEL LOGIC - VERSION 2
# ============================================================

def assign_better_label(row):

    best = row["best_score"]
    margin = row["score_margin"]
    cls = row["evidence_class"]

    # --------------------------------------------------------
    # 1. HIGH-CONFIDENCE CASE
    # --------------------------------------------------------
    # Strong evidence AND clear winner
    if best >= 0.65 and margin >= 0.20:
        return cls

    # --------------------------------------------------------
    # 2. MEDIUM-CONFIDENCE CASE
    # --------------------------------------------------------
    # Reasonable evidence, but not strong enough to automatically
    # call it final
    if best >= 0.50 and margin >= 0.15:
        return cls

    # --------------------------------------------------------
    # 3. AMBIGUOUS CASE
    # --------------------------------------------------------
    # Two classes have relatively similar evidence
    if best >= 0.40 and margin < 0.15:
        return "Needs_Review"

    # --------------------------------------------------------
    # 4. LOW-EVIDENCE CASE
    # --------------------------------------------------------
    # None of the three categories has meaningful evidence
    if best < 0.40:
        return "Other_Persistent_Thermal_Source"

    # Anything sitting between the above conditions
    return "Needs_Review"


df["provisional_label"] = df.apply(
    assign_better_label,
    axis=1
)
```
